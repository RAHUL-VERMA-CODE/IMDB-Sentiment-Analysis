```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
##sentences

sent=[
    'the glass of milk',
    'the cup of tea'
    ,'i am a good boy'
    ,'i am a good developer',
    'undersatnd the meaing of words ',
    'your videos are goods'
]
```

```python
# define the vocabulary size
voc_size=10000
```

```python
### one hot representation
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr ## gives index of words where 1 in sparse matrix
```

```python
## word embedding representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
 ## need to define the timesatmp but 
 # the sentences have diff no. of words so we choose max length
sent_length=8
# its add other words vector to 0(fixed size of all sentences)
embedded_docs=pad_sequences(one_hot_repr
                            ,padding='pre',
                            maxlen=sent_length)  

print(embedded_docs) 
```

```python
## feature representation
dim=10
```

```python
# model create
model=Sequential()
model.add(Embedding(input_dim=voc_size,output_dim=dim))
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
