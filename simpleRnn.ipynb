```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
```

```python
## load the imdb dataset

max_feature=10000 ## vocabulary size
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=max_feature)

## print the shape of the data
print(f'Training data shape:{x_train.shape},training lablels shape {y_train.shape}')
print(f'Testing data shape:{x_test.shape},testing labels shape{y_test.shape}')
```

```python
from tensorflow.keras.preprocessing import sequence

max_len = 500

x_train = sequence.pad_sequences(x_train, maxlen=max_len)
x_test = sequence.pad_sequences(x_test, maxlen=max_len)
x_train
```

# Train Simple RNN

```python
from tensorflow.keras.layers import Input
model=Sequential()
model.add(Input(shape=(max_len,)))
model.add(Embedding(max_feature,128))
model.add(SimpleRNN(128,activation='tanh'))
model.add(Dense(1,activation='sigmoid'))

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

```

```python
model.summary()
```

```python
## create an instance of earlystopping callback
from tensorflow.keras.callbacks import EarlyStopping
earlyStopping=EarlyStopping(monitor='val_loss',patience=2,restore_best_weights=True)
```

```python
history=model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    callbacks=[earlyStopping]
)
```

```python
model.save('simple_Rnn_model.h5')
```
